# Demo: Iris end-to-end
Este notebook demonstra preparação de dados, treino com `src.train`, avaliação e inferência com o modelo salvo.

```python
# 1) Importar bibliotecas
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from pathlib import Path
from sklearn import datasets

sns.set(style='whitegrid')
```

```python
# 2) Preparar dataset Iris e salvar como CSV
p = Path('data')
p.mkdir(exist_ok=True)
iris = datasets.load_iris(as_frame=True)
df = iris.frame
# adicionar coluna target com nomes (opcional)
df['target'] = iris.target
# salvar
csv_path = p / 'iris.csv'
df.to_csv(csv_path, index=False)
print('Saved to', csv_path)
df.head()
```

```python
# 3) Inspeção rápida
print('shape:', df.shape)
df.info()
df.describe()
df.sample(5, random_state=0)
```

```python
# 5) Transformações e feature engineering
from sklearn.preprocessing import StandardScaler
features = df.drop(columns=['target'])
scaler = StandardScaler()
features_scaled = scaler.fit_transform(features)

# converter de volta para DataFrame para inspeção
features_s_df = pd.DataFrame(features_scaled, columns=features.columns)
features_s_df.head()
```

```python
# 6) Visualizações básicas
plt.figure(figsize=(8,6))
sns.scatterplot(x='sepal length (cm)', y='sepal width (cm)', hue='target', data=df)
plt.title('Sepal length vs width (Iris)')
plt.show()

# pairplot
sns.pairplot(df, hue='target', vars=df.columns[:-1])
```
